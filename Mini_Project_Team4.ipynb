```python
#Creating the role category and grouping related role
def get_role_category(title):

    title = str(title).lower()

    if any(word in title for word in [
        'data scientist',
        'data analyst',
        'data engineer',
        'machine learning',
        'artificial intelligence',
        'ai engineer'
    ]):
        return 'Data and AI'

    elif any(word in title for word in [
        'software developer',
        'software engineer',
        'web developer',
        'java developer',
        'python developer'
    ]):
        return 'Software Development'

    elif any(word in title for word in [
        'sales',
        'business development'
    ]):
        return 'Sales'

    elif any(word in title for word in [
        'hr',
        'human resource',
        'recruiter'
    ]):
        return 'HR'

    elif any(word in title for word in [
        'finance',
        'accountant',
        'accounting'
    ]):
        return 'Finance'

    else:
        return 'Other'
```

```python
import pandas as pd

# 1. Create a separate dataframe containing only necessary columns and drop missing salaries
salary_work_df = df[
    ['work_mode', 'salary_midpoint_in_LPA']
].dropna().copy()

# 2. Function to identify and collect outliers per work mode group
def get_group_outliers(group):
    Q1 = group['salary_midpoint_in_LPA'].quantile(0.25)
    Q3 = group['salary_midpoint_in_LPA'].quantile(0.75)
    IQR = Q3 - Q1

    lower_fence = Q1 - 1.5 * IQR
    upper_fence = Q3 + 1.5 * IQR

    # Print the calculated fences for this specific work mode
    print(f"Work Mode: {group['work_mode'].iloc[0]}")
    print(f"  -> Lower Fence: {lower_fence:.2f} LPA")
    print(f"  -> Upper Fence: {upper_fence:.2f} LPA\n")

    # Filter rows that fall OUTSIDE the fences
    outliers = group[
        (group['salary_midpoint_in_LPA'] < lower_fence) |
        (group['salary_midpoint_in_LPA'] > upper_fence)
    ]
    return outliers

# 3. Print fences and extract all outliers across work modes
print("--- IQR Fences Per Work Mode ---")
all_outliers = salary_work_df.groupby('work_mode', group_keys=False).apply(get_group_outliers)

# 4. Preview the resulting dataframe containing only the outlier rows
print(f"Total outlier rows found across all work modes: {len(all_outliers)}")
print(all_outliers.head(10))
```

# Test 1 — Two-proportion Z-test

Business question

Does the proportion of fresher-friendly postings differ between tech and non-tech jobs?

Hypotheses
- \(H_0: p(tech)=p(non-tech))
- \(H_1: p(tech)!=p(non-tech))

This is a two-sided test because we want to know whether the proportions differ in either direction.

```python
from statsmodels.stats.proportion import proportions_ztest

tech = df[df['is_tech'] == True]
nontech = df[df['is_tech'] == False]

count = [
    tech['is_freshers_friendly'].sum(),
    nontech['is_freshers_friendly'].sum()
]

total = [len(tech), len(nontech)]

z_stat, p_value = proportions_ztest(count=count,nobs=total,alternative='two-sided')

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

# Test 2 — Two-Sample t-Test

Business question

Does the average disclosed salary differ between tech and non-tech postings?
We will compare only postings with a valid, positive disclosed salary midpoint and INR currency. This avoids comparing missing salaries or mixing currencies.
Hypotheses
- \(H_0: avgdisclosedsal(tech) = avgdisclosedsal(non-tech))
- \(H_1: avgdisclosedsal(tech) != avgdisclosedsal(non-tech))

```python
from statsmodels.stats.weightstats import ttest_ind, DescrStatsW, CompareMeans

salary_data = df[
    (df['salary_disclosed'] == True) &
    (df['salary_midpoint_in_LPA'] > 0) &
    (df['currency'] == 'INR')
]

tech_salary = salary_data[salary_data['is_tech'] == True]['salary_midpoint_in_LPA']

nontech_salary = salary_data[salary_data['is_tech'] == False]['salary_midpoint_in_LPA']

t_stat, p_value, dfree = ttest_ind(tech_salary,nontech_salary,

    alternative='two-sided'
)

mean_diff = tech_salary.mean() - nontech_salary.mean()



print("Tech mean salary:", tech_salary.mean())
print("Non-tech mean salary:", nontech_salary.mean())
print("Mean difference:", mean_diff)
# print("95% CI for difference:", ci)
print("t-statistic:", t_stat)
print("P-value:", p_value)
```

Conclusion and placement insight
Reject \(H_0\).

The average disclosed salary midpoint differs significantly between tech and non-tech postings. Tech postings have the higher observed mean.


For placement planning, technical roles may offer higher salary opportunities in this dataset. However, salary is not guaranteed, and the comparison includes only postings with disclosed, usable salaries. The medians also differ substantially (₹14 LPA versus ₹4 LPA), so the mean difference is not solely a useful description of every individual job.

# Test 3 — Chi-Square Test of Association

Business question

Is work mode (on-site, hybrid, or remote) associated with whether a posting is marked fresher-friendly?

Hypotheses

- \(H_0\): Work mode and fresher-friendly status are independent.

- \(H_1\): Work mode and fresher-friendly status are associated.

Both variables are categorical, so a chi-square test of association is appropriate.

```python
import numpy as np
import pandas as pd
from statsmodels.stats.contingency_tables import Table

table = pd.crosstab(
    df['work_mode'],
    df['is_freshers_friendly']
)

test = Table(table.values).test_nominal_association()

chi2_stat = test.statistic
p_value = test.pvalue

n = table.to_numpy().sum()


print(table)
print("Chi-square statistic:", chi2_stat)
print("Degrees of freedom:", test.df)
print("P-value:", p_value)
```

Conclusion and placement insight

Reject \(H_0\). Work mode and fresher-friendly status are statistically associated in this dataset.

# Test 5-One-Sample Proportion Z-Test


Business question:

Is the proportion of job postings marked as fresher-friendly significantly different from a benchmark of 25%?


Why this is useful: It gives us a simple, measurable insight into how much of the advertised job market is labelled as fresher-friendly, which is directly relevant to placement preparation.


Hypotheses
- H₀: The proportion of fresher-friendly postings is 25%.
- H₁: The proportion of fresher-friendly postings is different from 25%.
Use 95% confidence and \(alpha=0.05\).

```python
from statsmodels.stats.proportion import proportions_ztest

success=np.sum(df['is_freshers_friendly'])
total=len(df)

z_stat, p_value = proportions_ztest(count=success, nobs=total, value=0.25, alternative='two-sided')

print("Z-statistic:", z_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

Conclusion-The proportion of fresher-friendly postings is different from 25%.
